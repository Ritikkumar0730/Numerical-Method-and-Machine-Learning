```python
import numpy as np
import matplotlib.pyplot as plt


L = 2.0
Nx = 100
dx = L / Nx

# Define time step
dt = 0.001
T = 1.0
Nt = int(T / dt)


nu = 0.01
x = np.linspace(0, L, Nx)

u = np.sin(np.pi * x)

def spatial_derivative(u): # by central difference method
    dudx = np.zeros_like(u)
    dudx[1:-1] = (u[2:] - u[:-2]) / (2 * dx)
    dudx[0] = (u[1] - u[0]) / dx
    dudx[-1] = (u[-1] - u[-2]) / dx
    return dudx

def second_derivative(u): # by central difference method
    d2udx2 = np.zeros_like(u)
    d2udx2[1:-1] = (u[2:] - 2 * u[1:-1] + u[:-2]) / dx**2
    return d2udx2
def f(u):
    return -u * spatial_derivative(u) + nu * second_derivative(u)

def rk4_step(u, dt):
    k1 = dt * f(u)
    k2 = dt * f(u + k1 / 2)
    k3 = dt * f(u + k2 / 2)
    k4 = dt * f(u + k3)
    return u + (k1 + 2*k2 + 2*k3 + k4) / 6
for _ in range(Nt):
    u = rk4_step(u, dt)
plt.plot(x, u, label="Final Profile")
plt.xlabel("x")
plt.ylabel("u")
plt.title("Burgers' Equation - RK4 + VPMind")
plt.legend()
plt.show()
```
